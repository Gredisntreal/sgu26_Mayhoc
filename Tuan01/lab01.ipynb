```python
import torch
a = torch.randn(2, 3)
b = torch.randn(2, 3)
print("a:")
print(a)
print("b:")
print(b)

print("\na + b:")
print(a + b)
print("\na / b:")
print(a / b)
print("\na^2:")
print(a ** 2)
print("\nMatrix multiplication:")
print(a @ b.T)
```

### Phân biệt a*b và a @ b.T
Phép * nhân từng phân tử ở vị trí tương ứng của ma trân kết là một ma trận với các phần tử mới với shape không bị thay đổi 

<Br> Còn phép nhân ma trận @ đòi hỏi số cột của a phải bằng hàng của b vì shape của a(2,3) và b(2,3) nên không thỏa cần sử dụng .T để chuyển vị ma trận b thành shape (3,2)

```python
print("Shape a @ b.T")
a = torch.randn(2, 3)
b = torch.randn(2, 3)

result = a @ b.T
print(result.size())   # torch.Size([2, 2])
```

## Bài 8

```python
import torch
x = torch.tensor(
3.6,
requires_grad=True
)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

Đạo hàm của y=x^2 là dy = 2x dx
<br> => dy/dx = 2x
<br> Tại x=3.6 => 2 * 3.6 = 7.2

```python
manual_grad = 2 * 3.6
print("Dao ham tinh thu cong:", manual_grad) 
print("x.grad:", x.grad.item())   
print("Khop nhau:", abs(manual_grad - x.grad.item()) < 1e-6)
```

## Bài 9

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
X * X,
X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
```

### So sánh 3 giá trị với [1,2,3]
Hệ số X*X y[0] có sai số rất nhỏ .0100 so với kết quả 1.0100 so với w[0]=1, trước khi học hệ số ta thu được kết quả -0.6135 với sai số lớn, có thể thấy hệ số này hội tụ nhanh và chính xác nhất trong ba hệ sau khi học<br>

Hệ số X y[1] có hội tụ tương đối và vẫn còn sai lệch sau khi học hệ số nhưng đã cải thiện từ lệch gần 1 xuống còn lệch khoảng 0.2176 <br>

Hệ số hằng số y[2] có sai số vẫn còn rất lớn kể cả sau khi học hệ số vẫn chưa hội tụ<br>

### Vai trò
Nếu không có hàm optimizer.zero_grad() mỗi epoch khiến w.grad bị tích tụ gây sai giá trị dẫn đến hội tụ không đúng, hàm sẽ trả gradient của tham số về không để tính gradient mới mỗi epoch.<br>

mse.backward() dùng để ghi nhớ các phép toán đã thực hiện dạng đồ thị để tính kết quả dự đoán y.pred để lưu vào w.grad mà không cần trả trực tiếp.Định hướng để gradient giảm sai số<br>

hàm optimizer.step() dùng kết quả gradient vừa tính được để cập nhật giá trị thực tế dựa trên thuật đã chọn ở đây là nadam với lr=0.01


## Bài 10

```python
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

Các giá trị thỏa 4 phương trình đều áp ra các giá trị 9,1,8,2
<br>
Có thể có nhiều nghiệm do sau khi biến đổi có thể thấy y4= y1 - y3 + y2 nên bốn phương trình không độc lập tuyến tính
<br> Hệ có 4 ẩn nhưng chỉ có 3 rank nên nghiệm tự do là cả một đường thẳng, kết quả chỉ là một số nằm trên đường thẳng đó
<br> Sqerr có thể tiến về 0, gradient luôn hội tụ về một điểm trên đường thẳng nghiệm, dừng ở vị trí gần xuất phát với giá trị ngẫu nhiên , phụ thuộc vào cách tạo A,B,C,D không phải lõi của thuật toán
